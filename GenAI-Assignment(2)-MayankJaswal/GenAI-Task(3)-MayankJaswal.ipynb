## User menu (While loop + break/continue)

```python
"""Implement a simple whle loop driven menu that keeeps asking the user ti choose an action until they press q to quit
1 - Add order amount to a running list
2 - show all orders and totals after applying discounts
q - Quit
"""
user_option = 1
orders = [1200, 2500, 800, 1750, 3000]
while user_option != "":
    user_option = input("Please choose a option between 1,2 and q: ")
    if user_option == "1":
        adding_amount = int(input("Please add order amount: "))
        orders.append(adding_amount)
    elif user_option == "2":
        total_revenue = 0
        calculate_discount = 0
        for order_amount in orders:
            # Apply the following discount rules and print the final amount
            if order_amount >= 2000:
                calculate_discount = (15 / 100) * order_amount
            elif 1500 <= order_amount < 2000:
                discount = 10
                calculate_discount = (10 / 100) * order_amount
            elif 1000 <= order_amount < 1500:
                calculate_discount = (7 / 100) * order_amount
            final_amount = order_amount - calculate_discount
            # computing the total revenue after the discounts
            total_revenue += final_amount
        print("Your all orders list :", orders)
        print("Total revenue after the discount is :", total_revenue)
    elif user_option == "q":
        break
    else:
        continue

```
