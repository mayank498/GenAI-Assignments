### Task 1: Discount Rules

```python
# Write a program th;at reads an integer order_amount from the user using input()
try:
    order_amount = float(input("Enter your ordered amount: "))
    calculate_discount = 0
    fixed_tax = 5
    # Apply the following discount rules and print the final amount
    if order_amount >= 2000:
        print("Congratulations, Your are getting 15% of discount")
        calculate_discount = (15 / 100) * order_amount
    elif 1500 <= order_amount < 2000:
        print("Congratulations, Your are getting 10% of discount")
        calculate_discount = (10 / 100) * order_amount
    elif 1000 <= order_amount < 1500:
        print("Congratulations, Your are getting 7% of discount")
        calculate_discount = (7 / 100) * order_amount
    else:
        print("No Discount! shop more then 1000 to get discount")
    # Calculating the discounted amount from the order amount
    discounted_amount = order_amount - calculate_discount
    # Calculating and adding tax on discounted amount
    total_tax = (fixed_tax / 100) * discounted_amount
    final_amount = total_tax + discounted_amount
    print(
        f"Products price is {order_amount} and after discount {discounted_amount} And 5% of tax  {total_tax} And your final amount is {final_amount}"
    )
    print(calculate_discount)
except Exception as err:
    print("Please enter valid input OR Somthing went worng:", err)
```
