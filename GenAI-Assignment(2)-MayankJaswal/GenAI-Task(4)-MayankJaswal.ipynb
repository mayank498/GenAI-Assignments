## Task 4: Loop control with conditions(break and continue)

```python
# Given a list of daily sales: daily = [200,150,0,400,50,300], iterate through it with a for loop and:
    # if a day's value is -1, treat it as corrupted data and break the loop
    # if a day's value is 0, treat it as day with no sales and continue
    # for valid positive sales, add to total_sales and print the running total
    # print the final total after the loop completes(or stops due to currupted data)
daily= [200,150,0,400,50,-1,300]
total_sales =0
for daily_sale in daily:
    if daily_sale < 0 :
        print(daily_sale)
        break
    if daily_sale == 0:
        continue
    total_sales += daily_sale
print(total_sales)
```
