## Task 2: Process multiple orders(for loop)

```python
# Given a list of order amounts, use a for loop to apply the discount rules from task 1 to each order and print a summary table
orders = [1200, 2500, 800, 1750, 3000]
calculate_discount = 0
total_revenue = 0
count = 0
for order_amount in orders:
    discount = 0
    # Apply the following discount rules and print the final amount
    if order_amount >= 2000:
        discount = 15
        calculate_discount = (discount / 100) * order_amount
        count += 1
    elif 1500 <= order_amount < 2000:
        discount = 10
        calculate_discount = (discount / 100) * order_amount
        count += 1
    elif 1000 <= order_amount < 1500:
        discount = 7
        calculate_discount = (discount / 100) * order_amount
        count += 1
    else:
        print("No Discount! shop more then 1000 to get discount")
    # Calculating the discounted amount from the order amount
    final_amount = order_amount - calculate_discount
    # computing the total revenue after the discounts 
    total_revenue += final_amount
    print(
        f"Product price is {order_amount} and applied discount {discount}% And your final amount is {final_amount}"
    )
print("Total revenue after the discount is :", total_revenue)
print("Numbaer of products that received a discount:", count)
```
