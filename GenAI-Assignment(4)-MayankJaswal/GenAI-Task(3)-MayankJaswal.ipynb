# Task 3: Append new sales

```python
# Append these new sales to the same file:
sale = 5000
sales_content = open("sales_data.txt","a")
sales_content.write("\n"+str(sale))
sales_content.close()
```

```python
sales = [2500,1700]
sales_content = open("sales_data.txt", "a")
sales_content.write("".join("\n"+str(data) for data in sales))
sales_content.close()
```

```python
sales_content = open("sales_data.txt","r")
content = sales_content.readlines()
# Print the total number of lines after appending
print("Total lines of content:",len(content))
print(content)
sales_content.close()
```
