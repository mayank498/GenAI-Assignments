# Task 6: Read file safely(Error handling inside file handling only)

```python
import os
```

```python
# Ask the user for a filename to open
file_path = input("Please enter your file name:")
file_path = file_path+".txt"
```

```python
# if the file exists, read and print it
if os.path.exists(file_path):
    with open(file_path,'r') as file_content:
        content = file_content.read()
        print(content)
else:
    print(f"File not found. Please check the filename:{file_path}")
```
