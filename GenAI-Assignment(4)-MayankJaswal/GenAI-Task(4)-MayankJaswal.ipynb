# Task 4: Generate summary report from file

```python
# Read all sales values from sales_data.txt
with open("sales_data.txt","r") as sales_data:
    # reading and converting data into integers 
    content = [int(data) for data in sales_data.readlines()]
    # Calculating total sales
    total_sale = sum(content)
    # calculating highest sale
    highest_sale = max(content)
    # calculating lowest sale
    lowest_sale = min(content)
    # calculating average sale
    avg_sale = total_sale/len(content)
    print(content)
    print("total sales:", total_sale)
    print("highest sale is:", highest_sale)
    print("Lowest sale  is:", lowest_sale)
    print("average sale value is:", avg_sale)
```
