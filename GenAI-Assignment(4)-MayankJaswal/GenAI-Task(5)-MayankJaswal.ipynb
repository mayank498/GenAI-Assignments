# Task 5: Create product info file

```python
# Ask the user for 3 product names & their prices.
product = {}
for i in range(0,3):
    key = input("Enter Product name:")
    value = int(input("Enter the product price"))
    product[key]=value
```

```python
# Write them into a new file products.txt in this format: ProductName|Price
with open("products.txt","w") as products_content:
    products_content.writelines(f"{key} | {value}\n" for key, value in product.items())
```

```python
# Read the file and print each line with proper formatting
with open("products.txt","r") as product_and_value:
    content = product_and_value.read()
    print(content)
```
