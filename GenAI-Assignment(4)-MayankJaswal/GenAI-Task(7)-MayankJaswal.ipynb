# Task 7: Mini Project - Export discounted Prices

```python
# Create a dictionary:
prices = {
    "Mouse": 500,
    "Keyboard": 800,
    "Monitor": 7000,
    "Pendrive": 400,
    "Camera": 5000,
}
```

```python
discount_percentage = int(input("Enter the discount percentage:"))
discount_price={}
for key,value in prices.items():
    discount_price[key] = value - (discount_percentage/100)*prices[key]
print(discount_price)
```

```python
with open("discount_report.txt", "w") as report:
    report.write("Product | Orignal Price | Discount Price\n")
    report.writelines(f"{key} | {value} | {dis_value}\n" for (key, value),(_,dis_value) in zip(prices.items(), discount_price.items()))
```

```python
# After writing, read the file and print it to the terminal
with open("discount_report.txt","r") as report:
    product_report = report.read()
    print(product_report)
```

```python
# Write a summary at the bottom of the file:
with open("discount_report.txt", "r") as report:
    product_detials = report.readlines()
    total_items = len(product_detials)
    print(total_items)
    # counting  average discount
    total_sum = 0
    for _, value in discount_price.items():
        total_sum += value
    avg_discount = total_sum/len(discount_price)
    print(avg_discount)
```
