# Assignment 4: File Handling(Read, Write, Append, Modes)

## Task 1: write sales records to a file.

```python
# Create a list of sales amounts:
sales = [1200, 450, 980, 1500, 3000]
sales_data = open("sales_data.txt", "w")
sales_data.write("\n".join(str(sale) for sale in sales))
sales_data.close()
```

```python
# Close the file, then reopen it and print its contents
sales_content = open("sales_data.txt","r")
content = sales_content.read()
print(content)
sales_content.close()
```

```python
# Wrtie the data in comma-separated format instead of separatea lines
comma_separated_data = open("sales_data.txt", "w")
content = comma_separated_data.write(", ".join(str(sale) for sale in sales))
comma_separated_data.close()
```
