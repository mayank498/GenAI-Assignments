# Task 2: Read file in different ways

```python
# Read the entier file using .Read() and print it

sales_content = open("sales_data.txt","r")
content = sales_content.read()
print(content)
sales_content.close()
```

```python
# Read the first line using .readline().
sales_content = open("sales_data.txt","r")
content = sales_content.readline()
print(content)
sales_content.close()
```

```python
# Read all lines using .readlines() and convert them into a list of integers.
sales_content = open("sales_data.txt","r")
content = [int(data) for data in sales_content.readlines()]
print(content)
sales_content.close()
```
