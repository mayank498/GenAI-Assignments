## Task 6 - Combined Utility Function

```python
# calculate the percentage of any number
def calculate_percentage(number,percent):
    return (percent/100)*number
```

```python
# Write a function process_price that:
def process_price(prices):
    discount_applied_list = list(map(lambda price: price - calculate_percentage(price, 10), prices))
    filterd_list = list(filter(lambda price: price > 300, discount_applied_list))
    return discount_applied_list, filterd_list
```

```python
discount_applied_list, filterd_list = process_price([100,500,900,50,750])
print("10% Discount applied list: ",discount_applied_list)
print("Filtred discounted list above 300: ",filterd_list)
```
