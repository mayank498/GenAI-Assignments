### Task 4 - Using map(): Apply GST to list of Prices.

```python
# calculate the percentage of any number
def calculate_percentage(number,percent):
    return (percent/100)*number
```

```python
prices = [100, 250, 400, 1200, 50]
gst_percentage = 18
price_with_gst = list(map(lambda price: price+ calculate_percentage(price, gst_percentage),prices))
print("Orignal price of list:",prices)
print("List after applying GST on prices",price_with_gst)
```
