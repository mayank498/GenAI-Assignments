## Task 5 - Using filter(): Filter Expensive products.

```python
prices = [100,250,400,1200,50,2000,850]
```

```python
# Careate a list of prices greater than 500
graeater_list = list(filter(lambda x:x>500, prices))
lower_list = list(filter(lambda y:y<=500, prices))
print("Price list which is greater then 500:",graeater_list)
print("Price list which is lowe or equal to 500:",lower_list)
```
