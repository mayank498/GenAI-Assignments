## Task 1- Basic Function: Price after discount

```python
def calculate_items_discount(price, discount_percent):
    return (discount_percent/100)*price
```

```python
def apply_discount(price, discount_percent= 5):
    if discount_percent >= 60:
        return "Discount never exceeds 60%"
    return price - calculate_items_discount(price, discount_percent)
```

```python
print("Final product price after discount is:", apply_discount(1000,40))
```

```python
print("Final product price after discount is:", apply_discount(500))
```
