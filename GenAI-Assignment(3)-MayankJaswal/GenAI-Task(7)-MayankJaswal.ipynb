## Task 7 - Mini Problem: Menu using functions

```python
# Adding value in price list
def add_price(prices_list, price):
    return prices_list.append(price)
```

```python
# Get average price from price list
def get_average_price(prices_list):
    if len(prices_list) == 0:
        return 0
    return sum(prices_list)/len(prices_list)
     

```

```python
# Gen maximum price for the price list
def get_max_price(prices_list):
    if len(prices_list) == 0:
        return 0
    return max(prices_list)
```

```python
# Create a simple menu where user can add price and get maximum and average value also:
prices_list = []
option = input("Please select the options from a menu, 1: Add price, 2: show average price, 3: show s=highest price and q: Quit:")
while option != "":
    if option ==  "1":
        price = int(input("Enter price to add into a list:"))
        add_price(prices_list, price)
        print("Here is the updated price list: ", prices_list)
    elif option == "2":
        avg_price =get_average_price(prices_list)
        print("Average price from price list is: ", avg_price)
    elif option == "3":
        max_price = get_max_price(prices_list)
        print("Maximum price amount from price list is: ",max_price)
    elif option == "q":
        break
    else:
        print("Invalid value! Please enter a value from menu")
    option = input("Please select the options from a menu, 1: Add price, 2: show average price, 3: show s=highest price and q: Quit:")



```
