# Task 3 - Lambda function: GST Calculator

```python
# calculate the percentage of any number
def calculate_percentage(number,percent):
    return (percent/100)*number
```

```python
# Create a lambda function gst that returns price after adding 18% GST.
gst_percentage = 18
gst = lambda price: price+(calculate_percentage(price, gst_percentage))
```

```python
gst(2000)
```

```python
# calculate the price after applying the discount
def calculate_price_discount(price, discount):
    return price - calculate_percentage(price,discount) 
```

```python
# Create another lambda function to compute final price after GST+discount together.
discount = 15
final_price = lambda price: calculate_percentage(price, gst_percentage) + calculate_price_discount(price, discount)
```

```python
final_price(1000)
```
