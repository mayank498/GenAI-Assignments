## Task 2 - Recursive function: Factorial utility

```python
# Create a recursive function factorial(n)

def factorial(num):
    try:
        if num < 0 :
            return "Error: please enter positive integer"
        if num == 0 or num == 1:
            return 1
        else:
            return num*factorial(num-1)
    except ValueError as err:
        return err
```

```python
factorial(5)
```

```python
factorial(0)
```

```python
factorial(-3)
```
