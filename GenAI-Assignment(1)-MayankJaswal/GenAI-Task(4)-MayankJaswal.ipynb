```python
# Using the products list and price_dict, create a list of tuples named catalog where each tuple is (product_name, price, category).
catalog = [
    ("Laptop", 70000, "Electronics"),
    ("T-Shirt", 1200, "Clothing"),
    ("Apple", 150, "Fruits"),
    ("Keyboard", 2500, "Electronics"),
    ("Jeans", 2500, "Clothing"),
    ("Coffee", 450, "Home&kitchen"),
    ("Headphones", 5000, "Electronics")
]
```

```python
# From catalog, create a new dictonary category_to_products that maps each category to a list of product names in that category.
category_to_products = {
    "Electronics":["Laptop","Keyboard","Headphones"],
    "Clothing":["T-Shirt","Jeans"],
    "Fruits":["Apple"],
    "Home&kitchen":["Coffee"]
}
```

```python
# Print all products tha belong to the category that has the maximum number of products.
max_category = max(category_to_products, key=lambda x: len(category_to_products[x]))
print(max_category)
print(category_to_products[max_category])
```
