## Task 3: Product Pricing(Dictionaries)

```python
# Create a dictionary price_dict where keys are product names and values are prices.
price_dict = {
    "Headphones": 5000,
    "T-Shirt": 1200,
    "Air Fryer": 6000,
    "Green Tea": 400,
    "Dumbbells": 2500,
    "Gaming Mouse": 3000
}
print(price_dict)
```

```python
# Add a new product with price to  price_dict
price_dict["Face Wash"] = 330
#update the price of an existing product
price_dict["T-Shirt"] = 1500.50
```

```python
# Remove a product by name
del price_dict["Dumbbells"]
```

```python
print(price_dict)
```

```python
price_dict.pop("Dumbbells", None)
print(price_dict)
```

```python
# Print the average price of all products
products_price = price_dict.values()
avg_price = sum(products_price)/len(products_price)
print(avg_price)
```

```python
# Print the product with both the maximum and minimum prices.
print("Maximum pricing is:", max(price_dict.values()),", minimum pricing is:", min(price_dict.values()))
```
