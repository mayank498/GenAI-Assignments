## Product Collections(Lists and Tuples)

```python
# Create a list named products containing at least 6 product names.
products = ["Laptop", "T-shirt", "Gaming Mouse", "Sugar", "Air Fryer", "Dumbbell"]
```

```python
# Create a tuple names simple_product that stores(product_name, price, caregory) of one product.
sample_product = ("Laptop", 15000, "Electronics")
```

```python
# Print the 2nd and last product form the products list.
print("2nd product:",products[1], ",last product:", products[-1])
```

```python
# Append two new product names to products and then print the updated list.
products.append("Atomic Habits")
products.append("Face Wash")
```

```python
print("Updated products list:", products)
```

```python
# Convert simple_product into a list, change its price and convert it back to a tuple
sample_product = list(sample_product)
print("sample products list", sample_product)
sample_product[1] = 25000
print("Updataed sample products list", sample_product)
sample_product = tuple(sample_product)
print("Type of sample products", type(sample_product))
```
